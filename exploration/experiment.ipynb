```python
from pathlib import Path
import numpy as np
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

# From our earlier baseline notebook
SAMPLE_PATH = ROOT / "data/interim/v01_sample_24features.csv"

assert SAMPLE_PATH.exists(), f"Sample not found: {SAMPLE_PATH}"

df = pd.read_csv(SAMPLE_PATH, low_memory=False)

# Find the existing binary target column
target_options = [
    "target", "target_default", "default",
    "is_default", "loan_default", "label", "loan_status"
]

target_col = next(
    (c for c in target_options if c in df.columns),
    None
)

assert target_col is not None, "Target column not found."

if target_col == "loan_status":
    y = df[target_col].map({
        "Fully Paid": 0,
        "Charged Off": 1
    })
else:
    y = pd.to_numeric(df[target_col], errors="coerce")

assert y.notna().all(), "Missing or invalid target values found."
assert set(y.unique()) == {0, 1}, "Expected binary labels 0 and 1."

y = y.astype(int)

balance = y.value_counts().sort_index()

print("Dataset shape:", df.shape)
print("Target column:", target_col)
print("Class 0 (Fully Paid):", f"{balance[0]:,}")
print("Class 1 (Charged Off):", f"{balance[1]:,}")
print("Default rate:", f"{y.mean():.2%}")
print("Majority/minority ratio:", round(balance[0] / balance[1], 2))
```
